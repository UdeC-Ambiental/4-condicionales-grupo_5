# 🧩 Reto 1 – Temperatura
**Objetivo:** clasificar la temperatura del agua y evaluar posible estrés térmico.

**Datos:** `../data/muestras_calidad_agua.csv`

### Pasos
1. Carga el CSV en `df`.
2. Extrae `sitio` y `temperatura` en `df_temp`.
3. Crea `categoria_temp` con np.where o .loc.
4. Crea `riesgo_termico`.
5. Filtra a `df_caliente` (alta temperatura).
6. Exporta `resultado_temperatura.csv`.

> **Reto adicional:** porcentaje de sitios con riesgo térmico.

```python
import pandas as pd

# Paso 1: Cargar el archivo base de calidad del agua
df = pd.read_csv("../data/muestras_calidad_agua.csv")

# Verificación rápida de datos
print("Vista previa del archivo cargado:")
display(df.head())

# Paso 2: Extraer solo las columnas necesarias para este análisis
df_temp = df[["sitio", "temperatura (°C)"]].copy()

print("Datos seleccionados para el análisis térmico:")
display(df_temp)

# Paso 3: Clasificar cada sitio según su temperatura usando condicionales

def clasificar_temp(t):
    if t < 18:
        return "fría"
    elif t < 24:
        return "templada"
    elif t < 28:
        return "cálida"
    else:
        return "muy cálida"

df_temp["categoria_temp"] = df_temp["temperatura (°C)"].apply(clasificar_temp)

print("Clasificación térmica por sitio:")
display(df_temp)

# Paso 4: Estimar riesgo térmico según temperatura

def riesgo_termico(t):
    if t < 24:
        return "bajo"
    elif t < 28:
        return "medio"
    elif t <= 30:
        return "alto"
    else:
        return "crítico"

df_temp["riesgo_termico"] = df_temp["temperatura (°C)"].apply(riesgo_termico)

print("Riesgo térmico asignado:")
display(df_temp)

# Paso 5: Filtrar los sitios con temperatura alta o muy alta
df_caliente = df_temp[df_temp["riesgo_termico"].isin(["alto", "cítrico"])]

print("Sitios con posible estrés térmico:")
display(df_caliente)

# Paso 6: Exportar el archivo con resultados
df_temp.to_csv("../data/resultado_temperatura.csv", index=False)
print("Archivo exportado correctamente como 'resultado_temperatura.csv'")

# Reto adicional: porcentaje de sitios con riesgo medio, alto o crítico
total = len(df_temp)
riesgo = len(df_temp[df_temp["riesgo_termico"].isin(["medio", "alto", "cítrico"])])
porcentaje = (riesgo / total) * 100

print(f"Porcentaje de sitios con riesgo térmico: {porcentaje:.1f}%")

```
