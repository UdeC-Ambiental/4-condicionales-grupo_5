# 🧩 Reto 3 – Oxígeno disuelto
**Objetivo:** evaluar riesgo ecológico según oxígeno disuelto.

**Datos:** `../data/muestras_calidad_agua.csv`

### Pasos
1. Carga el CSV en `df`.
2. Extrae `sitio` y `oxigeno`.
3. Crea `nivel_oxigeno` (bajo/medio/alto).
4. Crea `riesgo_biologico` (⚠ si oxígeno < 5).
5. Filtra y exporta `resultado_oxigeno.csv`.

> **Reto adicional:** porcentaje de sitios con riesgo.

```python
# Escribe tu código aquí
import pandas as pd

df = pd.read_csv('/content/muestras_calidad_agua.csv')
display(df.head(10))
```

```python
sitio_oxigeno_10_df = df[['sitio', 'oxigeno']].head(10)
display(sitio_oxigeno_10_df)
```

```python
max_oxigeno = df['oxigeno'].max()
min_oxigeno = df['oxigeno'].min()
```

```python
max_oxigeno,min_oxigeno
```

```python
import numpy as np

def categorize_oxigeno(oxigeno):
    if oxigeno > 6:
        return 'Alto'
    elif oxigeno < 5:
        return 'Bajo'
    else:
        return 'Medio'

df['Nivel_Oxigeno'] = df['oxigeno'].apply(categorize_oxigeno)
display(df[['sitio', 'oxigeno', 'Nivel_Oxigeno']].head(10))
```

```python
df['riesgo_biológico'] = np.where(df['oxigeno'] < 5, 'Sí', 'No')
display(df[['sitio', 'oxigeno', 'riesgo_biológico']].head(10))
```

```python
# Filtrar por riesgo_biologico 'Sí'
filtered_df = df[df['riesgo_biologico'] == 'Sí']

# Seleccionar solo las columnas deseadas antes de exportar
filtered_df[['sitio', 'oxigeno', 'riesgo_biologico']].to_csv('resultado_oxigeno.csv', index=False)

print("DataFrame filtrado exportado a 'resultado_oxigeno.csv'")
```

```python
# Calcular el número total de sitios
total_sitios = len(df)

# Calcular el número de sitios con riesgo biologico 'Sí'
sitios_con_riesgo = len(df[df['riesgo_biologico'] == 'Sí'])

# Calcular el porcentaje
porcentaje_riesgo = (sitios_con_riesgo / total_sitios) * 100

print(f"El porcentaje de sitios con riesgo biologico es: {porcentaje_riesgo:.2f}%")
```
